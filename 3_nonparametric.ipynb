```python
import numpy as np
import matplotlib.pyplot as plt
from bokeh.plotting import figure, show, output_notebook
from bokeh.layouts import gridplot
```

```python
%matplotlib inline
output_notebook()
plt.style.use('dark_background')
```

```python
def weights_calculate(x0, X, tau):
    return np.exp(-np.sum((X - x0) ** 2, axis=1) / (2 * tau ** 2))
```

```python
def local_weighted_regression(x0, X, Y, tau):
    # Add bias term
    x0 = np.r_[1, x0]
    X_aug = np.c_[np.ones(len(X)), X]
    w = weights_calculate(x0, X_aug, tau)
    W = np.diag(w)
    XTWX = X_aug.T @ W @ X_aug
    XTWY = X_aug.T @ W @ Y
    theta = np.linalg.pinv(XTWX) @ XTWY
    return x0 @ theta
```

```python
def plot_lwr(tau):
    domain = np.linspace(-3, 3, num=300)
    prediction = [local_weighted_regression(x0, X, Y, tau) for x0 in domain]

    plot = figure(width=400, height=400)  # Updated attribute names
    plot.title.text = f'tau = {tau}'
    plot.scatter(X, Y, alpha=0.3)
    plot.line(domain, prediction, line_width=2, color='red')
    return plot
```

```python
n = 1000
X = np.linspace(-3, 3, num=n)
Y = np.abs(X ** 3 - 1)
X += np.random.normal(scale=0.1, size=n)
show(gridplot([
    [plot_lwr(10.0), plot_lwr(1.0)],
    [plot_lwr(0.1), plot_lwr(0.01)]
]))
```
