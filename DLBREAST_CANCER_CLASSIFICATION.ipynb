```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn.datasets
from sklearn.model_selection import train_test_split
```

```python
breast_datasets = sklearn.datasets.load_breast_cancer()
```

x and y


```python
x=df.drop(columns='label',axis=1)
y=df['label']
```

training and test

train_test_split

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
print(x.shape,x_train.shape,x_test.shape)
```

```python
#standardize the data
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
x_train_std=scaler.fit_transform(x_train)
x_test_std = scaler.transform(x_test)
```

```python
print(x_train_std)
```

Building neural network


6.import tensorflow and keras

```python
import tensorflow as tf
tf.random.set_seed(3)
from tensorflow import keras
```

```python
#setting up the layers of the model
model = keras.Sequential([  keras.layers.Flatten(input_shape=(30,)),
    keras.layers.Dense(20,activation='relu'),
    keras.layers.Dense(2,activation='sigmoid')

])
```

```python
#complie like setting temp of oven time etc
model.compile(optimizer="adam",
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
#fit check
history=model.fit(x_train_std,y_train,validation_split=0.1,epochs=10)
```

```python
#visualizing acuracy and loss
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title("model accuracy")
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.legend(['training data','validation data'],loc='lower right')
```

```python
# visualize loss
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title("model loss")
plt.xlabel('epoch')
plt.ylabel('loss')
plt.legend(['training data','validation data'],loc='upper right')
```

```python
#accracy of the model on test data
loss,accuracy = model.evaluate(x_test_std,y_test)
print(accuracy)
```
