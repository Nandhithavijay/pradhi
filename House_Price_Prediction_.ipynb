importing the dependencies


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn.datasets
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from sklearn import metrics
```

importing datasets


```python
house_dataset=sklearn.datasets.fetch_california_housing()
```

```python
print(house_dataset)
```

```python
house_dataframe = pd.DataFrame(house_dataset.data,columns = house_dataset.feature_names)
```

```python
house_dataframe['price']=house_dataset.target
```

checking the no row and columns in the dataframe


```python
print(house_dataframe.head())
```

```python
house_dataframe.isnull().sum()
```

```python
house_dataframe.shape
```

```python
house_dataframe.describe()
```

```python
correlation = house_dataframe.corr(numeric_only = True)
```

```python
plt.figure(figsize=(10,10))
sns.heatmap(correlation,cbar=True,square=True,fmt='.1f',annot=True,annot_kws={'size':8},cmap='Blues')
```

```python
#spliting the data and target
x = house_dataframe.drop(['price'],axis=1)
y = house_dataframe['price']
```

```python
print(x)
print(y)
```

```python
# to process training and testing datasets:splitting
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
print(x.shape,x_train.shape,x_test.shape)
```

```python
#model training:
#1.load model
model = XGBRegressor()
```

```python
#2.training the model
model.fit(x_train,y_train)
```

```python
#evaluation
#prediction on training data

#accuracy for prediction on training data
training_data_prediction = model.predict(x_train)


```

check for missing values:


```python
print(training_data_prediction)
```

```python
#r2_square
score1 = metrics.r2_score(y_train,training_data_prediction )
#mean absolute error
score2=metrics.mean_absolute_error(y_train,training_data_prediction)
```

```python
print("r2=",score1)
print("mean=",score2)
```

```python
test_data_prediction = model.predict(x_test)
```

```python
score3=metrics.r2_score(y_test,test_data_prediction)
score4=metrics.mean_absolute_error(y_test,test_data_prediction)
```

```python
print("r2=",score3)
print("mean=",score4)
```

```python
plt.scatter(y_train,training_data_prediction)
plt.xlabel("actual price")
plt.ylabel("predicted price")
plt.title("actual versus predicted")
plt.show()
```
